```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
%matplotlib inline
```

```python
df = pd.read_csv('Classified Data', index_col=0)
df.head()
```

```python
# Any variable value on a large scale will have a large influence on the distance. So we need to scale the data.
# When using KNN, it is always a good idea to scale the data. We can use StandardScaler from sklearn to do this.
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(df.drop('TARGET CLASS', axis=1))
```

```python
scaled_features = scaler.transform(df.drop('TARGET CLASS', axis=1))    # transform the data and convert it to an array
```

```python
scaled_features
```

```python
df_feat = pd.DataFrame(scaled_features, columns=df.columns[:-1])   # convert the array back to a dataframe with the same column names as the original dataframe
df_feat.head()
```

```python
from sklearn.model_selection import train_test_split
X = df_feat
y = df['TARGET CLASS']
X_train, X_test, y_train, y_test = train_test_split(scaled_features, df['TARGET CLASS'], test_size=0.30, random_state=101)
```

```python
from sklearn.neighbors import KNeighborsClassifier  # import the KNN model from sklearn.neighbors
knn = KNeighborsClassifier(n_neighbors=1)    # instantiate the model with n_neighbors=1
knn.fit(X_train, y_train)  # fit the model to the data
```

```python
pred = knn.predict(X_test)    # predict the labels for the test set
pred
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(confusion_matrix(y_test, pred))
print('*'*60)
print(classification_report(y_test, pred))
```

```python
error_rate = []

# Here we will use a loop to determine the error rate for K values between 1 and 40. We will then plot the error rate against K values to find the optimal K value.
for i in range(1, 40):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train, y_train)
    pred_i = knn.predict(X_test)
    error_rate.append(np.mean(pred_i != y_test))

error_rate
```

```python
plt.figure(figsize=(10,6))
plt.plot(range(1,40), error_rate, color='blue', linestyle='dashed', marker='o',markerfacecolor='red', markersize=10)
plt.title('Error Rate vs K-Value')
plt.xlabel('K')
plt.ylabel('Error Rate')
plt.show()
```

```python
# From the above diagram we see that around  k=17, the error rate is minimum. So we will choose k=17 as our KNN model and check the classification report and confusion matrix again.
knn = KNeighborsClassifier(n_neighbors=17)
knn.fit(X_train, y_train)
pred = knn.predict(X_test)
print(confusion_matrix(y_test, pred))
print('*'*60)
print(classification_report(y_test, pred))
```

```python
# From above output with K = 17, we see that the error rate has reduced and the classification report is better than the previous one with K=1. So we can conclude that KNN with K=17 is a better model than KNN with K=1 for this dataset.
```
