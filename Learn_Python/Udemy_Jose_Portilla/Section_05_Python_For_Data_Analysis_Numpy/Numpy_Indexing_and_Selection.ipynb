# Numpy Indexing and Selection

```python
import numpy as np
```

```python
arr = np.arange(0,11)
arr
```

```python
print(arr[8])
print(arr[1:5])
print(arr[0:5])
print(arr[:6])
print(arr[5:])
print(arr[4:8])
```

```python
arr[0:5] = 100
arr
```

```python
arr =np.arange(0,11)
arr
```

```python
slice_of_arr = arr[0:6]
slice_of_arr
```

```python
slice_of_arr[:] = 99
slice_of_arr
```

```python
# original array also changed because slice_of_arr is a view of arr not a copy
arr
```

```python
arr_copy = arr.copy()
arr_copy
```

```python
arr_copy[:] = 100
arr_copy
```

```python
# Indexing a 2D array (matrix)
arr_2d = np.array([[5,10,15],[20,25,30],[35,40,45]])
arr_2d
```

```python
print(arr_2d[0]) # first row
print(arr_2d[1]) # second row
print(arr_2d[2]) # third row

print(arr_2d[0][0]) # first row, first column
print(arr_2d[0][1]) # first row, second column
print(arr_2d[1][1]) # second row, second column
print(arr_2d[2][0]) # third row, first column
print(arr_2d[2,2]) # third row, third column
```

```python
# we can use , instead of [][] to access elements in a 2D array

print(arr_2d[0,0]) # first row, first column
print(arr_2d[0,1]) # first row, second column
print(arr_2d[1,1]) # second row, second column
print(arr_2d[2,0]) # third row, first column
print(arr_2d[2,2]) # third row, third column
```

```python
# return 10,15 and 25,30
print(arr_2d[:2,1:])
```

```python
print(arr_2d[:1])
print(arr_2d[:2])
print(arr_2d[:3])
```

```python
print(arr_2d[:3,2:])
```

```python
print(arr_2d[:3,:])
```

```python
# print last 2 rows
print(arr_2d[1:3,:])
```
