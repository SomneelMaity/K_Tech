# Map

```python
# The map() function in Python is a built-in tool that applies a specific function to every item 
# in an iterable (like a list, tuple, or dictionary) and returns an iterator called a map object
# map(function, iterable, ...)

def times2(var):
    return var*2

seq = [1,2,3,4,5]
list(map(times2, seq))
```

```python
t = lambda var: var*2
t(4)
```

```python
list(map(lambda var: var*3, seq))
```

```python
# filter
# The filter() function in Python is a built-in tool used to extract elements from an iterable 
# (like a list, tuple, or set) that satisfy a specific condition

list(filter(lambda item: item%2 == 0, seq))
```
