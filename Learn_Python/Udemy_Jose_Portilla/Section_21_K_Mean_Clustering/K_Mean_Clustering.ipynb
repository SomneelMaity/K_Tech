```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
%matplotlib inline
```

```python
from sklearn.datasets import make_blobs
data = make_blobs(n_samples=200, n_features=2, centers=4, cluster_std=1.8, random_state=101)
data
```

```python
data[0]
```

```python
data[0].shape
```

```python
plt.scatter(data[0][:,0], data[0][:,1])     # data[0][:,0] is the first column of data[0] and data[0][:,1] is the second column of data[0]
```

```python
plt.scatter(data[0][:,0], data[0][:,1], c=data[1], cmap='rainbow')     # data[1] is the cluster labels for each point in data[0]
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=4)
kmeans.fit(data[0])
```

```python
kmeans.cluster_centers_
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10,6), sharey=True)   # sharey=True means that the y-axis will be the same for both plots
ax1.set_title('K Means')    # set the title of the first plot to 'K Means'
ax1.scatter(data[0][:,0], data[0][:,1], c=kmeans.labels_, cmap='rainbow')   # kmeans.labels_ is the cluster labels for each point in data[0] as determined by the KMeans algorithm

ax2.set_title('Original')   # set the title of the second plot to 'Original'
ax2.scatter(data[0][:,0], data[0][:,1], c=data[1], cmap='rainbow')  # data[1] is the cluster labels for each point in data[0] as determined by the make_blobs function
```
