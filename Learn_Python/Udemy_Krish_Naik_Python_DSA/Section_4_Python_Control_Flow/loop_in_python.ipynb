```python
## Practical Example 4: Find prime numbers
# Check if a number is prime

num = 17
is_prime = True

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    
    if is_prime:
        print(f"{num} is a prime number")
    else:
        print(f"{num} is not a prime number")
else:
    print(f"{num} is not a prime number")
```

```python
## Practical Example: Data Filtering with Continue
# Calculate average of positive numbers only

numbers = [10, -5, 20, -3, 15, 0, 8]
total = 0
count = 0

for num in numbers:
    if num <= 0:
        continue  # Skip non-positive numbers
    total += num
    count += 1

average = total / count if count > 0 else 0
print(f"Average of positive numbers: {average}")  # 13.25
```

```python
## Break vs Continue vs Pass Comparison
print("=== Break: Exit the loop ===")
for i in range(5):
    if i == 3:
        break  # Stops loop completely
    print(i)  # Prints: 0, 1, 2

print("\n=== Continue: Skip current iteration ===")
for i in range(5):
    if i == 3:
        continue  # Skips only i=3
    print(i)  # Prints: 0, 1, 2, 4

print("\n=== Pass: Do nothing ===")
for i in range(5):
    if i == 3:
        pass  # Does nothing, continues normally
    print(i)  # Prints: 0, 1, 2, 3, 4
```

```python
## Break Statement - Advanced Example
# Search for an element in nested lists

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
target = 5
found = False

for row in matrix:
    for element in row:
        if element == target:
            print(f"Found {target} in the matrix!")
            found = True
            break  # Exit inner loop
    if found:
        break  # Exit outer loop

if not found:
    print(f"{target} not found")
```

```python
## While vs Do-While Comparison
print("=== Regular While Loop ===")
x = 5
while x < 3:  # Condition false, never executes
    print("This won't print")
print("While loop completed\n")

print("=== Do-While Simulation ===")
x = 5
while True:
    print(f"This prints at least once! x = {x}")  # Executes once
    if not (x < 3):  # Then checks condition
        break
print("Do-while simulation completed")
```

```python
## Do-While Example: Menu System
# Execute at least once, then ask if user wants to continue

choice = ""
while True:
    print("Menu:")
    print("1. Option 1")
    print("2. Option 2")
    print("3. Exit")
    
    choice = "3"  # Simulating user choice
    print(f"You selected: {choice}")
    
    if choice == "3":
        print("Exiting...")
        break
    
# This executes at least once, like do-while
```

```python
## Do-While Loop Simulation - Method 2
# Using a flag variable

count = 0
first_iteration = True

while first_iteration or count < 3:
    print(f"Count: {count}")
    count += 1
    first_iteration = False
# Output: Count: 0, 1, 2
```

```python
## Do-While Loop Simulation in Python
# Python doesn't have do-while, but we can simulate it
# Do-while executes at least once, then checks condition

# Method 1: Using while True with break
count = 0
while True:
    print(f"Executing: count = {count}")
    count += 1
    if not (count < 3):  # Condition check at the end
        break
# Output: Executing: count = 0, 1, 2
```

```python
## While Loop - Infinite loop with break
# Use break to exit infinite loop

count = 0
while True:
    print(f"Count: {count}")
    count += 1
    if count >= 5:
        print("Breaking out of infinite loop")
        break
# Output: Count: 0, 1, 2, 3, 4
```

```python
## While Loop with user input simulation
# Keep asking until correct answer

password = "python123"
user_input = ""
attempts = 0

while user_input != password and attempts < 3:
    user_input = "wrong"  # Simulating user input
    attempts += 1
    if user_input == password:
        print("Access granted!")
    else:
        print(f"Wrong password! Attempt {attempts}/3")

if attempts == 3 and user_input != password:
    print("Account locked!")
```
