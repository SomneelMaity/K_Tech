```python
## Conditional Statements
# if statement - executes code block if condition is True
age = 18
if age >= 18:
    print("You are an adult")  # You are an adult
```

```python
## Real-world example: Temperature checker
temperature = 30

if temperature > 35:
    print("It's very hot!")
elif temperature > 25:
    print("It's warm")  # It's warm
elif temperature > 15:
    print("It's pleasant")
elif temperature > 5:
    print("It's cold")
else:
    print("It's freezing!")
```

```python
## Leap Year Checker
# A year is a leap year if:
# 1. It is divisible by 4 AND
# 2. If divisible by 100, then it must also be divisible by 400

year = 2024

if year % 400 == 0:
    print(f"{year} is a leap year")  # 2024 is a leap year
elif year % 100 == 0:
    print(f"{year} is not a leap year")
elif year % 4 == 0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")

# Test with different years
year = 2000
if year % 400 == 0:
    print(f"{year} is a leap year")  # 2000 is a leap year
elif year % 100 == 0:
    print(f"{year} is not a leap year")
elif year % 4 == 0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")

year = 1900
if year % 400 == 0:
    print(f"{year} is a leap year")
elif year % 100 == 0:
    print(f"{year} is not a leap year")  # 1900 is not a leap year
elif year % 4 == 0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```

```python
## Simple Calculator
num1 = 10
num2 = 5
operation = "+"

if operation == "+":
    result = num1 + num2
    print(f"{num1} + {num2} = {result}")  # 10 + 5 = 15
elif operation == "-":
    result = num1 - num2
    print(f"{num1} - {num2} = {result}")
elif operation == "*":
    result = num1 * num2
    print(f"{num1} * {num2} = {result}")
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
        print(f"{num1} / {num2} = {result}")
    else:
        print("Error: Division by zero!")
else:
    print("Invalid operation! Use +, -, *, or /")

# Test with different operations
operation = "*"
if operation == "+":
    result = num1 + num2
    print(f"{num1} + {num2} = {result}")
elif operation == "-":
    result = num1 - num2
    print(f"{num1} - {num2} = {result}")
elif operation == "*":
    result = num1 * num2
    print(f"{num1} * {num2} = {result}")  # 10 * 5 = 50
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
        print(f"{num1} / {num2} = {result}")
    else:
        print("Error: Division by zero!")
else:
    print("Invalid operation! Use +, -, *, or /")
```

```python
## Ternary operator (conditional expression)
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)  # Adult

# Another example
num = 15
result = "Even" if num % 2 == 0 else "Odd"
print(result)  # Odd
```

```python
## Multiple conditions with logical operators
age = 25
income = 50000

if age >= 18 and income >= 30000:
    print("Eligible for credit card")  # Eligible for credit card
else:
    print("Not eligible")
```

```python
## Nested if statements
age = 20
has_license = True

if age >= 18:
    if has_license:
        print("You can drive")  # You can drive
    else:
        print("You need a license")
else:
    print("You are too young to drive")
```

```python
## if-elif-else statement
marks = 75

if marks >= 90:
    print("Grade: A")
elif marks >= 80:
    print("Grade: B")
elif marks >= 70:
    print("Grade: C")  # Grade: C
elif marks >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```
