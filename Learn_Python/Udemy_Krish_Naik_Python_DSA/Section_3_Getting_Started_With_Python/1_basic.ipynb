```python
## Basic Syntax Rules in Python
## Case Sensitivity - Python is case sensitive. For example, the variable name "myVariable" is different from "myvariable".

name = "Neel"
Name = "Maity"
print(name)  # Output: Neel
print(Name)  # Output: Maity
```

```python
## Indentation - Python uses indentation to define blocks of code. Indentation is crucial for defining the structure of your code.
age = 18
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
print(age)  # Output: 18
```

```python
## Line Continuation - Python allows you to break long lines of code into multiple lines for better readability. 
# You can use the backslash (\) character to indicate that a line continues on the next line.
total = 10 + 20 + 30 + \
        40 + 50
print(total)  # Output: 150
```

```python
## Multiple statements on a single line
x=5; y=10; z=15
print(x, y, z)  # Output: 5 10 15
```

```python
## Understand Semantics in Python
## Variable Assingment - In Python, you can assign values to variables using the assignment operator (=). The variable name should follow certain rules, such as starting with a letter or underscore and not containing spaces or special characters.
age = 32
print(type(age))
print(type(name))
```

```python
## Type Inference - Python is a dynamically typed language, which means that you don't need to explicitly declare the type of a variable. The interpreter infers the type based on the assigned value.
x = 10  # x is inferred as an integer
print(x)
print(type(x))  # Output: <class 'int'>
x = "Somneel" # x is now inferred as a string
print(x)
print(type(x))  # Output: <class 'str'>
```

```python
## Understanding Variable types
## Python is dynamically types, type of variable is determined at runtime.

## Type Conversion - Python allows you to convert between different data types using built-in functions like int(), float(), str(), etc.
age_str = "30"
age_int = int(age_str)
print("Age (string):", age_str)
print("Age (integer):", age_int)

Age = 30
_age = str(Age)
print("Age (integer):", Age)
print("Age (string):", _age)
```

```python
## Dynamic Typing
var = 10  # var is an integer
print(var, type(var))  # Output: 10 <class 'int'>
var = "Hello"  # var is now a string
print(var, type(var))  # Output: Hello <class 'str'>
var = 3.14  # var is now a float
print(var, type(var))  # Output: 3.14 <class 'float'>
```

```python
## input() function - The input() function allows you to take user input from the console. It returns the input as a string.
age = input("What is your age? ")
print(age, type(age))  # Output: (user input) <class 'str'>
```

```python
age = int(input("What is your age? "))
print(age, type(age))  # Output: (user input) <class 'int'>
```
