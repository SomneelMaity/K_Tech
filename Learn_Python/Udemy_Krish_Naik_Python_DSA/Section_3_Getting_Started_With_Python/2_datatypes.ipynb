```python
## Arithmatic Operators
a = 10
b = 3

# Addition
print(a + b)  # 13

# Subtraction
print(a - b)  # 7

# Multiplication
print(a * b)  # 30

# Division (float result)
print(a / b)  # 3.3333333333333335

# Floor Division (integer result)
print(a // b)  # 3

# Modulus (remainder)
print(a % b)  # 1

# Exponentiation (power)
print(a ** b)  # 1000
```

```python
## Comparison Operator
a = 10
b = 5

# Equal to
print(a == b)  # False

# Not equal to
print(a != b)  # True

# Greater than
print(a > b)  # True

# Less than
print(a < b)  # False

# Greater than or equal to
print(a >= b)  # True

# Less than or equal to
print(a <= b)  # False

# Comparison with same values
print(10 == 10)  # True
print(10 != 10)  # False
```

```python
## Logical Operator
# and - Returns True if both conditions are True
print(True and True)    # True
print(True and False)   # False
print(False and False)  # False

# or - Returns True if at least one condition is True
print(True or False)    # True
print(False or False)   # False
print(True or True)     # True

# not - Returns opposite boolean value
print(not True)         # False
print(not False)        # True

# Combining logical operators
a = 10
b = 5
c = 15

print(a > b and b < c)  # True (both conditions are True)
print(a > b or b > c)   # True (first condition is True)
print(not(a > b))       # False (a > b is True, not makes it False)

# Complex example
age = 25
has_license = True
print(age >= 18 and has_license)  # True (eligible to drive)
```
