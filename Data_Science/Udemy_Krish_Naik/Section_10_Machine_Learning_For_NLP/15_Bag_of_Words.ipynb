```python
import pandas as pd
messages = pd.read_csv("spam.csv", encoding='latin-1')
# Keep only first 2 columns and rename them
messages = messages.iloc[:, :2]
messages.columns = ['label', 'message']
```

```python
messages
```

```python
## Data Cleaning and Preprocessing
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
corpus = []
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-z]', ' ', str(messages['message'][i]))   ## Replace non-alphabetic characters with space
    review = review.lower()  ## Convert to lower case
    review = review.split()  ## Split the review into individual words
    review = [ps.stem(word) for word in review if not word in stopwords.words('english')]  ## Remove stop words
    review = ' '.join(review)  ## Join the words back into a single string
    corpus.append(review)  ## Add the cleaned review to the corpus
```
