```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
## PorterStemmer
## It is a rule-based stemmer. It is a simple and fast stemming algorithm that works by applying 
# a set of rules to the words. It is not very accurate, but it is very fast and can be used for 
# large datasets.
```

```python
from nltk.stem import PorterStemmer
```

```python
porter_stemmer = PorterStemmer()
```

```python
for word in words:
    print(word + " --> " + porter_stemmer.stem(word))
```

```python
porter_stemmer.stem('congratulation')
```

```python
## RegexpStemmer Class
## NLTK has RegexpStemmer class with the help of which we can easily implement 
# Regular Expression Stemmer algorithms. It basically takes a single regular expression and 
# removes any prefix or suffix that matches the expression. Let us see an example
```

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)      
## Here we are removing the suffixes 'ing', 's', 'e' and 'able' from the words. 
# The min parameter is used to specify the minimum length of the word after stemming. 
# If the length of the word after stemming is less than the min parameter, then the word is not stemmed.
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
reg_stemmer = RegexpStemmer('ing|s$|e$|able$', min=4)
reg_stemmer.stem('ingeating')
```

```python
### Snowball Stemmer
## It is a stemming algorithm which is also known as the Porter2 stemming algorithm as it is a 
# better version of the Porter Stemmer since some issues of it were fixed in this stemmer.
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(f"{word} --> {porter_stemmer.stem(word)}")
```

```python
## Difference in Snowball Stemmer and Porter Stemmer
porter_stemmer.stem('fairly'), porter_stemmer.stem('sportingly')
```

```python
snowballstemmer.stem('fairly'), snowballstemmer.stem('sportingly')
```
