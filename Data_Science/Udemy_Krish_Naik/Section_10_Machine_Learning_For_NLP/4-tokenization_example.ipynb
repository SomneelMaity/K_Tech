```python
corpus="""Hello Welcome,to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

```python
##  Tokenization
## Sentence-->paragraphs
from nltk.tokenize import sent_tokenize, word_tokenize      ## sent_tokenize-->sentence tokenization, word_tokenize-->word tokenization
```

```python
import nltk
nltk.download('punkt_tab')      ## Downloading the punkt_tab tokenizer model required for sent_tokenize
```

```python
documents = sent_tokenize(corpus)      ## sentence tokenization
print(documents)
print(type(documents))      ## type of documents is list
for sentences in documents:
    print(sentences)
```

```python
## Tokenization 
## Paragraph-->words
## sentence--->words

words = word_tokenize(corpus)      ## word tokenization
print(words)
print(type(words))      ## type of words is list
for word in words:
    print(word)
```

```python
from nltk.tokenize import wordpunct_tokenize      ## wordpunct_tokenize-->word punctuation tokenization
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer     ## TreebankWordTokenizer-->Treebank word tokenizer
```

```python
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
## Here full stop(.) is not considered as a separate token. It is attached to the word before it. 
## But for the last sentence, the full stop is considered as a separate token. This is because the last sentence is not followed by any other word.
```
