```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
np.random.seed(100) # generate 100 random data points
x = np.random.rand(50, 1) * 100 # random x values between 0 and 100
y = 3.5 * x + np.random.randn(50, 1) * 20 # linear relationship with some noise
```

```python
# training the linear regression model
model = LinearRegression()
model.fit(x, y)
```

```python
y_pred = model.predict(x) # this will give us the predicted y values based on our model
```

```python
# plotting the original data points and the regression line
plt.scatter(x, y, color='blue', label='Data Points')
plt.plot(x, y_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.title('Linear Regression on Random Data')
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Coefficient (slope):", model.coef_[0][0])
print("Intercept:", model.intercept_[0])
```
